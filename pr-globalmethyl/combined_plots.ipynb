```python
from pathlib import Path

OUT_DIR = Path("combined_plots_forFigPanels")
OUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# ================================
# Config - PNG files
# ================================

plot_files = {
    "chrom1k":    "globalplots_per_feature_chrom1k/line_density_hist_chrom1k_CHN.png",
    "chrom10k":   "globalplots_per_feature_chrom10k/line_density_hist_chrom10k_CHN.png",
    "chrom100k":  "globalplots_per_feature_chrom100k/line_density_hist_chrom100k_CHN.png",
    "genebody":   "globalplots_per_feature_genebody/line_density_hist_genebody_CHN.png",
    "promoter":   "globalplots_per_feature_promoter/line_density_hist_promoter_CHN.png",
}

titles = {
    "chrom1k":    "1kb Windows",
    "chrom10k":   "10kb Windows", 
    "chrom100k":  "100kb Windows",
    "genebody":   "Gene Bodies",
    "promoter":   "Promoters",
}

feature_order = ["chrom1k", "chrom10k", "chrom100k", "genebody", "promoter"]

# ================================
# Create combined figure
# ================================

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes_flat = axes.flatten()
axes_flat[5].set_visible(False)  # Hide 6th panel

for i, feat in enumerate(feature_order):
    ax = axes_flat[i]
    img = mpimg.imread(plot_files[feat])
    ax.imshow(img)
    ax.set_title(titles[feat], fontsize=13, fontweight='bold', pad=10)
    ax.axis('off')

fig.suptitle('Methylation Distribution Shifts Across Genomic Scales (mCH)', 
             fontsize=16, fontweight='bold', y=0.98)

plt.tight_layout()

plt.savefig(OUT_DIR / "panel_b_combined_mCH.pdf", dpi=300, bbox_inches='tight')
plt.savefig(OUT_DIR / "panel_b_combined_mCH.png", dpi=300, bbox_inches='tight')

plt.show()
print("✓ Saved combined figure")
```
